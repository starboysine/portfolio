```python
import requests

url = "https://api.open-meteo.com/v1/forecast"
params = {
    "latitude": 53.383,
    "longitude": -1.467,
    "current_weather": True
}
response = requests.get(url, params=params)
print("Status Code: ", response.status_code)

data = response.json()
print(data)

```

```python
print("Current_temperature: ", data["current_weather"]["temperature"], " Celsius")
```
